```python
words_columns=["brand", "fuel" ,"seller_type" ,"transmission" ,"owner"]
```

```python
num_columns=["year" ,"km_driven" ]
```

```python
print("wordscolum", words_columns)
print("numcolumns", num_columns)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
preprocessor= ColumnTransformer(transformers=[("with words",OneHotEncoder(handle_unknown="ignore"),words_columns),"numcolums","passthrough",num_columns])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test =train_test_split(x,y,test_size=0.2 , random_state=42)
```

```python
print("x_train shape", x_train.shape)
print("x_test shape", x_test.shape)
print("y_train shape", y_train.shape)
print("y_test shape", y_test.shape)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("with words", OneHotEncoder(handle_unknown="ignore"), words_columns),
        ("numcolums", "passthrough", num_columns)
    ]
)
x_train_encoded = preprocessor.fit_transform(x_train)
```

```python
x_test_encoded = preprocessor.transform(x_test)
```

```python
print("Rows after cleaning:", len(df))
```

```python
print("Training shape:", x_train_encoded.shape)
```

```python
print("Training prices:", y_train.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(x_train_encoded, y_train)
```

```python
y_pred = model.predict(x_test_encoded)
```

```python
print(y_pred[:5])
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(comparison.head(8))
```
